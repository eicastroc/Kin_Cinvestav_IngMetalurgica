# 1. Introducción

## 1.1 Generalidades

Reacciones heterogéneas, en las que un fluido (gas o líquido) contacta un sólido, reacciona con el mismo, y lo transforma en un producto. Este tipo de reacciones se pueden representar esquemáticamente como:


$$\begin{matrix*}[l] \mathrm{A}_{\text{fluido}} + \mathrm{bB}_{\text{sólido}} & \rightarrow & \text{productos fluidos} \\  & \rightarrow & \text{productos sólidos} \\ & \rightarrow & \text{productos fluidos y sólidos}\end{matrix*}

La tasa de cambio de la cantidad de especie $\mathrm{B}$ es una función compleja del tamaño de partícula, superficie de intercambio, fuerza motriz, coeficientes de transferencia de masa, coeficientes de difusividad efectiva, cinética de reacción, etc.

$$\frac{dN_{\mathrm{B}}}{dt} = f\left(r, R, S_{\mathrm{ex}}, \underbrace{(C_{\mathrm{A,g}} - C_{\mathrm{A,s}})}_{\text{fuerza motríz}}, k_{\mathrm{g}}, D_{e}, r_\mathrm{AB}, \cdots\right)$$

## 1.2 Modelos

Debajo se esquematizan dos casos de comportamiento de la partícula sólida al reaccionar con el fluido.

- En el primer caso, la partícula mantiene su tamaño a lo largo de la reacción. A medida que la reacción procede, y se consume reactivo, su volumen original disminuye y es progresivamente ocupado por los productos de reacción. Esta fenomenología se describe con el modelo de nucleo decreciente.

- En el segundo caso, la partícula disminuye en tamaño a medida que va reaccionando. Los productos de reacción se desprenden a medida que la reacción progresa. Esta fenomenología se describe con el modelo de partícula decreciente.

## 1.3 Consideraciones

### 1.3.1 Leyes cinéticas

El sistema de interés es heterogéneo, e involucra la reacción entre un fluido y un sólido. Las expresiones convenientes para la tasa de reacción química se resumen en la ***Tabla 1***, debajo.

<center>

***Tabla 1. Definiciones para la tasa de reacción convenientes para sistemas fluido-sólido.***

| Reacción | Tipo de sistema | Normalización | Tasa de reacción |
|---------|------------------|-------|:----------------:|
| Heterogénea | entre un fluido y un sólido | por unidad de masa del sólido | $$r_i^{\prime} = \frac{1}{W} \frac{dN_i}{dt} \quad \quad [\mathrm{mol.kg^{-1}.s^{-1}}]$$ |
| Heterogénea | gas - sólido | por unidad de superficie del sólido | $$r_i^{\prime\prime} = \frac{1}{S} \frac{dN_i}{dt} \quad \quad [\mathrm{mol.m^{-2}.s^{-1}}]$$ | 
| Heterogénea | gas - sólido | por unidad de volumen del sólido | $$r_i^{\prime\prime\prime} = \frac{1}{V_s} \frac{dN_i}{dt} \quad \quad [\mathrm{mol.m^{-3}.s^{-1}}]$$ |

</center>

Se reitera que la selección de la expresión de tasa de reacción es más que nada una selección de la expresión que sea más conveniente para el sistema estudiado. Las distintas expresiones de tasa de reacción presentadas en la ***Tabla 1*** están relacionadas entres si, ya que:

$$W r_i^{\prime} = S r_i^{\prime\prime} = V_s r_i^{\prime\prime\prime}$$

Considerando el caso de consumo del reactivo $i$, y el uso de modelos cinéticos tipo serie de potencias, estos se escriben:

$$r_{i}^{\prime} = -k^{\prime} \prod_{j=1}^{k} C_{j}^{n_j} \quad | \quad C_j [\mathrm{mol.kg^{-1}}]$$

$$r_{i}^{\prime \prime} = -k^{\prime \prime} \prod_{j=1}^{k} C_{j}^{n_j} \quad | \quad C_j [\mathrm{mol.m^{-2}}]$$

$$r_{i}^{\prime \prime \prime} = -k^{\prime \prime \prime} \prod_{j=1}^{k} C_{j}^{n_j} \quad | \quad C_j [\mathrm{mol.m^{-3}}]$$

De forma general, las dimensiones de la constante cinética deben tener dimensiones de forma:


$$k \left[\mathrm{C^{1 -\sum n_j}.t^{-1}}\right]$$

<details>
<summary><b>Nomenclatura</b></summary>

<div class="alert alert-info">

- **$k$**: constante cinética con dimensiones apropiadas,  $[\mathrm{C^{1 -\sum n_j}.t^{-1}}]$.

- **$C$**: Concentración con unidades apropiadas, $[\mathrm{mol.kg^{-1}}], \, [\mathrm{mol.m^{-2}}], \, [\mathrm{mol.m^{-3}}]$.

- **$\sum n_j$**: orden de la reacción, $[-]$.

- **$t$**: tiempo en unidades apropiadas, $[\mathrm{s}], \, [\mathrm{min}], \, \cdots$

</div>

</details>

### 1.3.2 Difusión de especies químicas

El modelado de los fenómenos de interés requiere de información sobre la difusividad efectiva de especies químicas en la capa límite entre fluido y partícula, y/o al interior de la partícula.

La difusividad efectiva es una medida de la capacidad de una especie para difundirse a través de un medio no homogéneo y poroso. De forma simplificada se puede representar como una función de la difusividad molecular de la especie en el fluido ($D$), la porosidad del material ($\epsilon$), y su tortuosidad ($\tau$).

$$D_{e} = D \frac{\epsilon}{\tau}$$

Modelos más complejos toman en cuenta la distribución de los poros en el medio, la temperatura, etc.

### 1.3.3 Transferencia de masa entre un fluido y un sólido

El intercambio de especies químicas entre un fluido y un sólido ocurre a través de la capa límite en la superficie de la partícula,  que delimita estas dos regiones. 

La resistencia de la capa límite depende de muchos factores tales como la velocidad relativa entre la partícula y el fluido, el tamaño de la partícula, propiedades termofísicas de la partícula y el fluido, etc.



En la literatura existen correlaciones adimensionales, que consideran varias situaciones de contacto entre fluidos y sólidos. Por ejemplo, lechos empacados, lechos fluidizados, sólidos en caida libre, etc.

La correlación de Froessling relaciona la transferencia de masa de una especie con fracción molar $y$ en un fluido hacia esferas en caida libre:

$$\mathrm{Sh} = 2.0 + 0.6 \left(\mathrm{Re}\right)^{\frac{1}{2}} \left(\mathrm{Sc}\right)^{\frac{1}{3}}$$

$$\left(\frac{k_g d_p y}{D} \right) = 2.0 + 0.6 \left( \frac{\rho d_p u}{\mu} \right)^{\frac{1}{2}} \left( \frac{\mu}{\rho D} \right)^{\frac{1}{3}}$$

<details>
<summary><b>Nomenclatura</b></summary>

<div class="alert alert-info">

- **$\mathrm{Sh}$**: número de Sherwood.

- **$\mathrm{Re}$**: número de Reynolds.

- **$\mathrm{Sc}$**: número de Schmidt.

- **$k_g$**: coeficiente de transferencia de masa, $[\mathrm{m.s^{-1}}]$.

- **$d_p$**: diámetro de partícula, $[\mathrm{m}]$

- **$y$**: fracción molar de especie química de interés en el fluido.

- **$D$**: difusividad de la especie química de interés en el fluido, $[\mathrm{m^{2}.s^{-1}}]$.

- **$\rho$**: densidad del fluido, $[\mathrm{kg.m^{-3}}]$.

- **$u$**: velocidad relativa entre el fluido y la partícula, $[\mathrm{m.s^{-1}}]$.

- **$\mu$**: Viscosidad dinámica del fluido, $[\mathrm{Pa.s}] = [\mathrm{kg.m^{-1}.s^{-1}}]$.

</div>

</details>

```python
Re = np.logspace(0, 4, 50)
Sc = np.logspace(0, 3, 50)

def Sherwood(Re, Sc):
    return 2.0 + 0.6 * np.power(Re,1/2) * np.power(Sc, 1/3)

RE, SC = np.meshgrid(Re, Sc)
SH = Sherwood(RE, SC)

fig, ax = plt.subplots()

cs = ax.contour(RE, SC, SH)
custom_fmt = lambda x: f"Sh={x:.0f}"
ax.clabel(cs, inline=True, fmt=custom_fmt)
ax.set_xlabel(r"$\mathrm{Re}$")
ax.set_ylabel(r"$\mathrm{Sc}$")
ax.set_xlim(0, 1e4)
ax.set_ylim(0, 1e3)
ax.grid(ls='--')
plt.show()
```
