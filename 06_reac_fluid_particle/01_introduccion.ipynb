# 1. Introducción

## 1.1 Consideraciones

Reacciones heterogéneas, en las que un fluido (gas o líquido) contacta un sólido, reacciona con el mismo, y lo transforma en un producto. Este tipo de reacciones se pueden representar esquemáticamente como:


$$\begin{matrix*}[l] \mathrm{A}_{\text{fluido}} + \mathrm{bB}_{\text{sólido}} & \rightarrow & \text{productos fluidos} \\  & \rightarrow & \text{productos sólidos} \\ & \rightarrow & \text{productos fluidos y sólidos}\end{matrix*}

La tasa de cambio de la cantidad de especie $\mathrm{B}$ es una función compleja del tamaño de partícula, superficie de intercambio, fuerza motriz, coeficientes de transferencia de masa, coeficientes de difusividad efectiva, cinética de reacción, etc.

$$\frac{dN_{\mathrm{B}}}{dt} = f\left(r, R, S_{\mathrm{ex}}, \underbrace{(C_{\mathrm{A,g}} - C_{\mathrm{A,s}})}_{\text{fuerza motríz}}, k_{\mathrm{g}}, D_{e}, r_\mathrm{AB}, \cdots\right)$$

## 1.2 Modelos

Debajo se esquematizan dos casos de comportamiento de la partícula sólida al reaccionar con el fluido.

- En el primer caso, la partícula mantiene su tamaño a lo largo de la reacción. A medida que la reacción procede, y se consume reactivo, su volumen original disminuye y es progresivamente ocupado por los productos de reacción. Esta fenomenología se describe con el modelo de nucleo decreciente.

- En el segundo caso, la partícula disminuye en tamaño a medida que va reaccionando. Los productos de reacción se desprenden a medida que la reacción progresa. Esta fenomenología se describe con el modelo de partícula decreciente.
