# E 1

$$\mathrm{CH_4 + O_2 \rightarrow CO_2 + H_2O}$$

**1. Definir sistema de ecuaciones para el balance:**

$$\mathrm{\nu_1 CH_4 + \nu_2 O_2 \rightarrow \nu_3 CO_2 + \nu_4 H_2O}$$


- Escribir el balance estequiométrico para cada elemento:

<center>

| Especie | Balance estequiométrico |
|--------|----------|
| $\mathrm{C}$ | $-\nu_1 + \nu_3 = 0$ |
| $\mathrm{H}$  | $-4\nu_1 + 2\nu_4 = 0$ |
| $\mathrm{O}$  | $-2\nu_2 +2\nu_3 + \nu_4 = 0$ |

</center>

Escribir el sistema de ecuaciones:

```python
# escritura de la matriz en forma escalonada
A = np.matrix([[-1, 0, 1, 0],
               [-4, 0, 0, 2],
               [0, -2, 2, 1]])
A = sp.Matrix(A)
A
```

**2. Resolver el sistema de ecuaciones lineales:**

```python
# obtener forma escalonada reducida
Arref = A.rref()
Arref[0]
```

Si:

$$\nu_r = 2$$

Entonces:

$$\nu_1 = 1$$

$$\nu_2 = 2$$

$$\nu_3 = 1$$

$$\nu_4 = 2$$


```python
nr = sp.sympify(2)
sp.Matrix([-nr* Arref[0].col(-1), nr])
```

**3. Escribir la ecuación balanceada:**

$$\mathrm{CH_4 + 2 O_2 \rightarrow  CO_2 + 2 H_2O}$$
